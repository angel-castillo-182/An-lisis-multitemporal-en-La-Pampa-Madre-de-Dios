```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
#
```

```python
df_tmax = pd.read_csv('df_tmax_corregido.csv')
df_tmax.head()
```

```python
fig, axes = plt.subplots( figsize=(8, 3))

df_tmax['valor'].plot(ax=axes)
```

```python
df_tmax['fecha'] = pd.to_datetime(df_tmax['fecha'])
df_tmax.fecha
```

```python
df_2000 = df_tmax[df_tmax['fecha'].dt.year == 2000]
df_2010 = df_tmax[df_tmax['fecha'].dt.year == 2010]
```

```python
fig, axes = plt.subplots(nrows= 2, ncols=1, figsize=(8, 3), constrained_layout =True )

df_2000.set_index('fecha')['valor'].plot(ax=axes[0],color='k')
df_2010.set_index('fecha')['valor'].plot(ax=axes[1],color='r')

axes[0].set_xlabel('')
```

```python
df_tmax['mes'] = df_tmax['fecha'].dt.month
df_tmax.head()
```

```python
fig, axes = plt.subplots( figsize=(8, 3), constrained_layout = True)

df_tmax.boxplot(ax=axes, column='valor', by = 'mes')
```
